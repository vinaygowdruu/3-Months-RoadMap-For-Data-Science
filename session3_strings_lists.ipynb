### Sequence sum

1/1! + 2/2! + 3/3! + ...

```python
# Code here

n = int(input('enter n \t'))

result = 0
fact = 1

for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact

print(result)

#Why n+1?

#range(1, 5+1) = range(1, 6) → produces [1, 2, 3, 4, 5] ✓

#range(1, 5) → produces [1, 2, 3, 4] ✗ (missing 5)

```

```python
# code here

rows = int(input('enter number of rows'))

for i in range(1,rows+1):
  for j in range(1,i+1):
    print('*',end='')
  print()
```

**Alternative Code to the above**



```python
rows = int(input('enter number of rows'))
symbol="*"
for i in range(1, rows + 1):
  print(symbol * i)
```

### Pattern 2

1234321<br>

```python
# Code here
rows = int(input('enter number of rows'))

for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j,end='')
  for k in range(i-1,0,-1):
    print(k,end='')

  print()
```

# Loop Control Statement

# - **Break**
   

*   Search in a database/list	Stop searching once you find the matching record instead of scanning all 10,000 rows
*   ATM PIN verification	Stop after 3 wrong attempts and lock the card

# - **Continue**



*   Data cleaning	Skip rows with missing/invalid data in a CSV file, process the rest

*   Email validation	Skip empty lines or malformed emails in a contact list


# **- Pass**
"Implement this feature later" without breaking the code


```python
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower,upper+1):
  for j in range(2,i):
    if i%j == 0:
      break
  else:
    print(i)
```

```python
# Continue
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```

```python
for i in range(1,10):
  pass
```
